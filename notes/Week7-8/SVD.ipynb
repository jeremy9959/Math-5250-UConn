```python
import scipy
import numpy as np
import matplotlib.pyplot as plt
```

```python
A=scipy.datasets.ascent()
```

```python
U, s, Vt = np.linalg.svd(A, full_matrices=False)
```

```python
plt.imshow(A,cmap="gray")
```

```python
plt.plot(np.arange(1, len(s) + 1), s, color="C0")
plt.xlabel("index $i$")
plt.ylabel(r"singular value $\sigma_i$")
plt.title("Singular values of the ascent image")
plt.show()
```

```python
def Xk(A,k):
    U, s, Vt = np.linalg.svd(A, full_matrices=False)
    Sk = np.diag(s[:k])
    return U[:, :k] @ Sk @ Vt[:k, :]
```

```python
plt.imshow(Xk(A,5),cmap="gray")
```

```python
plt.imshow(Xk(A,20),cmap="gray")
```
